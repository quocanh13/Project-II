```python
import pandas as pd
import random
# =====================================================================
# 1. TẬP HỢP CÁC KỊCH BẢN TẤN CÔNG BEC & SOCIAL ENGINEERING
# =====================================================================
# Kịch bản 1: Sếp nhờ vả gấp (Gift cards / Wire transfer)
exec_greetings = ["Are you at your desk?", "Quick favor needed.", "Are you available right now?", "I need your assistance urgently.", "Can you handle a quick task for me?"]
exec_contexts = ["I am currently in a board meeting and cannot take calls.", "I'm traveling and my signal is terrible.", "I am stepping into a client presentation.", "I'm tied up with the auditors all morning.", "I am out of the office and need this handled discreetly."]
exec_requests = ["I need you to purchase five $100 Apple gift cards for a client reward program.", "Please process a wire transfer of $14,500 to our new vendor.", "Go to the nearest store and buy 10 physical Amazon gift cards, scratch them, and email me the codes.", "I need an urgent outbound payment routed to the account details I will provide next.", "Can you authorize a quick expense payment? I will send the invoice details shortly."]
exec_constraints = ["Do not call me, just reply to this email when it's done.", "This is a surprise, so keep it strictly confidential.", "I need this completed in the next 30 minutes.", "Please confirm once you have read this so I know you are on it.", "Use your corporate card and I will approve the reimbursement by EOD."]
exec_signoffs = ["Sent from my iPhone", "Regards,", "Thx,", "Best,", "- Sent via mobile device"]
# Kịch bản 2: Lừa đảo đổi số tài khoản ngân hàng đối tác (Vendor Invoice Fraud)
vendor_greetings = ["Hi Accounts Payable,", "Good morning Finance team,", "Hello,", "To the accounting department,", "Dear Billing Contact,"]
vendor_contexts = ["Our company is currently undergoing an internal financial audit.", "We recently transitioned to a new banking institution.", "Our previous bank account is temporarily frozen due to a technical issue.", "We are updating our financial records for the upcoming tax season.", "Due to restructuring, our payment routing has changed."]
vendor_requests = ["Please update our banking details for all future invoices to the new routing number below.", "Ensure the pending payment for invoice #4912 is redirected to our new escrow account.", "Attached is our updated W-9 and new ACH wiring instructions. Please update your system.", "We request that you hold the current check and issue a wire transfer to our new corporate account instead.", "Can you confirm the status of our last invoice? Also, note our new banking details for the remittance."]
vendor_constraints = ["Please confirm receipt of this update.", "Do not send payments to the old account as they will bounce.", "We need this updated in your system before Friday.", "Let me know if you need any additional forms to process this change.", "Effective immediately, all remittances must go to the new account."]
# Kịch bản 3: Lừa đảo tiền lương nhân sự (Payroll/HR Fraud)
hr_greetings = ["Hi HR,", "Hello Payroll team,", "Hi,", "To Human Resources,", "Good afternoon,"]
hr_contexts = ["I just changed my bank and need to update my direct deposit.", "My old checking account was compromised this weekend.", "I am switching to a new credit union for better rates.", "I lost my debit card and had to close my primary account.", "I need to reroute my salary for this upcoming pay cycle."]
hr_requests = ["Can you please change my direct deposit information to the new account?", "I have attached a voided check for my new account. Please update my payroll profile.", "What is the process to update my bank details before Friday's payroll run?", "Please halt my current deposit and use the new routing and account numbers I provide below.", "Can you send me the form to change my direct deposit account immediately?"]
hr_constraints = ["I need this effective for the current pay period to avoid late fees.", "Please let me know when this is done.", "I am locked out of the employee portal, so I need you to do it manually.", "Let me know if there's any issue processing this today.", "Thanks for handling this so quickly."]
# Kịch bản 4: Tấn công phi kỹ thuật IT (IT Spoofing - No links, just reply)
it_contexts = ["We are finalizing the migration to the new Microsoft 365 server.", "An unusual login attempt was detected on your corporate email.", "Your password is set to expire in 2 hours.", "We are auditing inactive accounts to free up license seats.", "A recent security patch requires manual synchronization."]
it_requests = ["Please reply to this email with your current password to verify your session.", "To prevent an account lockout, send us your employee ID and current login PIN.", "Reply immediately with 'CONFIRM' and your current password to keep your account active.", "We need you to authenticate manually. Please reply with your username and password.", "Send your current credentials to this secure IT thread to complete the update."]
it_constraints = ["Failure to do so will result in an immediate network lockout.", "This is a mandatory security protocol.", "Do not share this request with other departments.", "Please complete this before logging off today.", "Your immediate cooperation is required."]
# =====================================================================
# 2. THUẬT TOÁN SINH DỮ LIỆU ĐA DẠNG
# =====================================================================
generated_emails = set()
def generate_combinations(greet_list, context_list, req_list, constraint_list, signoff_list=None):
   for _ in range(500): # Thử tạo thật nhiều để lọc set
       greet = random.choice(greet_list)
       context = random.choice(context_list)
       req = random.choice(req_list)
       constraint = random.choice(constraint_list)
       if signoff_list:
           signoff = random.choice(signoff_list)
           email = f"{greet} {context} {req} {constraint} {signoff}"
       else:
           email = f"{greet} {context} {req} {constraint}"
       generated_emails.add(email)
# Trộn từ tất cả các kịch bản
generate_combinations(exec_greetings, exec_contexts, exec_requests, exec_constraints, exec_signoffs)
generate_combinations(vendor_greetings, vendor_contexts, vendor_requests, vendor_constraints)
generate_combinations(hr_greetings, hr_contexts, hr_requests, hr_constraints)
generate_combinations(["IT Helpdesk:", "System Admin:", "Security Team:"], it_contexts, it_requests, it_constraints)
# Rút trích chính xác 500 mẫu độc nhất
final_bec_list = random.sample(list(generated_emails), 500)
# Tạo DataFrame
df_bec = pd.DataFrame({
   'text': final_bec_list,
   'label': 1  # Tất cả đều là Spam (Phishing/BEC)
})
# Lưu ra file CSV
output_file = './src/500_BEC_Spam_Dataset.csv'
df_bec.to_csv(output_file, index=False, quoting=1)
print(f"🎉 Đã tạo thành công {len(df_bec)} email lừa đảo BEC/Social Engineering siêu thực tế!")
print(f"📁 File đã được lưu tại: {output_file}")
# In thử 3 câu ngẫu nhiên để kiểm tra văn phong
print("\n🔍 XEM THỬ 3 MẪU BEC NGẪU NHIÊN:")
for i, text in enumerate(df_bec['text'].sample(3)):
   print(f"\n{i+1}. {text}")
```
